### Scatter 

- A scatter plot is a two-dimensional data visualization that shows distinct data points as dots or markers. The scatter plot's dots are arranged according to their values along the X and Y axes, and each one represents a single data observation. In order to visualize correlations between two numerical variables and spot patterns, clusters, or trends in the data, scatter plots are extremely helpful.


- The plt.scatter() tool provided by Matplotlib makes it simple to build scatter plots.

```python
import matplotlib.pyplot as plt

# Sample data
x = [1,2,3,4,5,7,8,9]
y = [12,15,18,12,9,15,19,12]

# Create a scatter plot
plt.scatter(x, y, color='y', marker='o')

# Add labels and title
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Basic Scatter Plot")

# Display the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns  # Seaborn for data loading and better styling
import pandas as pd

# Load the Iris dataset from Seaborn
iris = sns.load_dataset("iris")

# Calculate the average sepal length for each species
average_sepal_length = iris.groupby("species")["sepal_length"].mean()

# Convert the result to a Pandas DataFrame for easier plotting
average_sepal_length_df = pd.DataFrame(average_sepal_length).reset_index()

# Create the bar plot
plt.figure(figsize=(8, 6))  # Adjust the figure size
plt.bar(average_sepal_length_df["species"], average_sepal_length_df["sepal_length"], color='y')
plt.xlabel("Species")
plt.ylabel("Average Sepal Length")
plt.title("Average Sepal Length for Each Species")
plt.show()
```

### Histogram

- A histogram is a type of data visualization that shows how a continuous or numerical dataset is distributed. It counts the number of data points that fall into each bin after dividing the range of values into intervals known as "bins". The resulting graphic sheds light on the data's fundamental distribution by exhibiting characteristics like peaks, valleys, skewness, and gaps.


- Histograms can be easily made with Matplotlib's plt.hist() function.

```python
import matplotlib.pyplot as plt
import numpy as np

# Generate random data for the histogram
data = np.random.normal(0, 1, 1000)

# Create a basic histogram
plt.hist(data, bins=20, color='y', edgecolor='g')
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram Plot")
plt.show()
```

#### stacked histogram >> two histogram together 

```python
import matplotlib.pyplot as plt
import numpy as np

# Generate random data for two distributions
data1 = np.random.normal(0, 1, 1000)
data2 = np.random.normal(3, 1, 1000)

# Create a stacked histogram
plt.hist([data1, data2], bins=20, color=['skyblue', 'salmon'], edgecolor='black', label=['Distribution 1', 'Distribution 2'])
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Stacked Histogram")
plt.legend()
plt.show()
```

#### Customized Histogram

```python
import matplotlib.pyplot as plt
import numpy as np

# Generate random data for the histogram
data = np.random.gamma(2, 1, 400)

# Create a customized histogram with additional options
plt.hist(data, bins=20, color='green', edgecolor='black', alpha=0.8, density=True)
plt.axvline(data.mean(), color='red', linestyle='dashed', linewidth=1, label='Mean')
plt.xlabel("Value")
plt.ylabel("Probability Density")
plt.title("Customized Histogram")
plt.legend()
plt.show()
```

### What is Donut Chart ? 

+ A Donut Chart (also spelled doughnut chart) is a variant of a pie chart, with a blank center hole giving it the shape of a donut. It is a circular statistical graphic divided into slices to illustrate numerical proportion.

```python
import matplotlib.pyplot as plt

# 1. Define the data
categories = ['Data Science', 'Machine Learning', 'Deep Learning', 'Web Deployment', 'Databases']
data_values = [30, 25, 20, 15, 10]
colors = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99', '#c2c2f0']

# 2. Create the figure and axis
fig, ax = plt.subplots(figsize=(6, 6))

# 3. Draw the pie chart (the foundation of the donut chart)
wedges, texts, autotexts = ax.pie(
    data_values, 
    labels=categories, 
    autopct='%1.1f%%', 
    startangle=90, 
    colors=colors,
    pctdistance=0.85, # Place percentages closer to the outer edge
    textprops=dict(color="black") # Label text color
)

# 4. Draw a white circle at the center to create the "donut hole"
centre_circle = plt.Circle((0,0), 0.70, fc='white')
fig.gca().add_artist(centre_circle)

# 5. Customize text inside the slices for better readability
for autotext in autotexts:
    autotext.set_color('darkblue')
    autotext.set_fontsize(10)
    autotext.set_weight('bold')

# 6. Add a title or total value inside the center hole
ax.text(0, 0, 'Technical\nStack', ha='center', va='center', fontsize=14, weight='bold', color='#333333')

# 7. Final adjustments and display
ax.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
plt.title('Course Distribution Overview', fontsize=16, pad=20, weight='bold')
plt.tight_layout()
plt.show()
```

### what is Box-plot ?

+ A Box Plot (also known as a box-and-whisker plot) is a standardized, non-parametric statistical graphic used to display the distribution of a continuous numerical dataset based on a five-number summary. It visually abstracts the data's central tendency, symmetry, skewness, and variance while instantly highlighting extreme values.

### What is violin plot ?

+ A Violin Plot is an advanced statistical graphic that merges the structural metrics of a box plot with a rotated, mirrored Kernel Density Plot on each side. While a box plot only reveals summary statistics, a violin plot shows the entire continuous probability density of the data at different values.

```python
plt.violinplot(data['Balance'])
```

### what is Area plot 


area plot >> variables changed over

 
+ time An area plot, also called a stacked area plot, is a type of data visualization that shows how various quantitative factors have changed over time while highlighting how they all work together to form the whole

+ It's particularly good for illustrating how elements of a whole change over time or along a continuous dimension

+  A layered representation of the data is produced in an area plot by layering many variables on top of one another.

+ The cumulative influence of all levels provides insight into the overall trend. Each layer represents a separate category or variable

+ To distinguish between categories, the space beneath each layer is frequently colored.


```python
years = [2021, 2018, 2022, 2023, 2017]
category1 = [20, 25, 30, 28, 35]
category2 = [10, 15, 20, 18, 25]
category3 = [5, 8, 10, 9, 15]

plt.stackplot(years, category1, category2, category3, labels = ["category1", "category2", "category3",],
             colors = ["skyblue","lightgreen", "orange"])
plt.xlabel("years")
plt.ylabel("Values")
plt.title("Area plot")
plt.legend()
plt.show()
```
